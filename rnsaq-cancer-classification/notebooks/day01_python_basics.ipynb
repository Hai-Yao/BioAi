```python
seq = "ATGCGATCGATCGTTAGCGGCTAACG"
print(seq)
#excercise 1: Count the number of A, T, G, and C in the sequence
count_A = seq.count('A')
count_T = seq.count('T')
count_G = seq.count('G')
count_C = seq.count('C')
# print the counts
print("Number of A:", count_A)
print("Number of T:", count_T)
print("Number of G:", count_G)
print("Number of C:", count_C) 
```

```python
#excercise 2: Calculate the GC content of the sequence
gc_content = (count_G + count_C) / len(seq) * 100
print("GC content: {:.2f}%".format(gc_content))

#excercise 3: write a function to calculate the GC content of a given sequence
def gc_content(seq):
    count_G = seq.count('G')
    count_C = seq.count('C')
    gc_content = (count_G + count_C) / len(seq) * 100
    return gc_content

print("GC content of the sequence: {:.2f}%".format(gc_content(seq)))
```

```python
#exercise 5: find motif
seq = "ATGCGATGACCATGGCTA"
#find the position of the motif "ATG" in the sequence
motif = "ATG"
positions = []
for i in range(len(seq) - len(motif) + 1):
    if seq[i:i+len(motif)] == motif:
        positions.append(i)
print("Positions of motif 'ATG':", positions)

#find the codens in the sequence
codons = [seq[i:i+3] for i in range(0, len(seq), 3)]
print("Codons in the sequence:", codons)

#find the position of the ATG codons in the sequence
atg_positions = []
for i in range(len(seq) - 2):
    if seq[i:i+3] == "ATG":
        atg_positions.append(i)
print("Positions of ATG codons:", atg_positions) 
```

```python
#exercise 6: reverse complement of the sequence
def reverse_complement(seq):
    complement = {'A': 'T', 'T': 'A', 'G': 'C', 'C': 'G'}
    rev_comp = ''.join(complement[nucleotide] for nucleotide in reversed(seq))
    return rev_comp
print("Reverse complement of the sequence:", reverse_complement("ATGC"))
```

```python
#excercise 7:gene expression dictionary
expression = {
    "TP53": 8.2,
    "EGFR": 22.1,
    "BRCA1": 5.4,
    "KRAS": 13.7,
    "MYC": 18.5,
    "PTEN": 3.1
}

highly_expressed = []
for gene, value in expression.items():
    if value > 10:
        highly_expressed.append(gene)
print("Highly expressed genes:", highly_expressed)

#find the highest expressed gene
highest_gene = max(expression, key=expression.get)
print("Highest expressed gene:", highest_gene, "with expression level:", expression[highest_gene])

#sorting the genes by expression level
sorted_genes = sorted(expression.items(), key=lambda x: x[1], reverse=True)
print("Genes sorted by expression level:")
for gene, value in sorted_genes:
    print(gene, ":", value)
```

```python
#conprehensive questions

np.random.seed(123)

X = np.random.rand(80, 500) #80 samples, 500 genes

labels = np.array(
    ["control"] * 40 +
    ["disease"] * 40
)

#calculate the number of all samples
num_samples = X.shape[0]
X.shape[0]
print("Number of samples:", num_samples)

#calculate the number of all genes
num_genes = X.shape[1]
print("Number of genes:", num_genes)

mean_expression_per_gene = X.mean(axis=0)
#print("Mean expression level per gene:", mean_expression_per_gene)
variance_expression_per_gene = X.var(axis=0)
# print("Variance of expression level per gene:", variance_expression_per_gene)
top_20_variable_genes_indices = np.argsort(variance_expression_per_gene)[-20:]
top_20_variable_genes_names = [f"Gene_{i}" for i in top_20_variable_genes_indices]
print("Top 20 variable genes names:", top_20_variable_genes_names)

disease_x = X[labels == "disease"]
healthy_x = X[labels == "control"]
mean_expression_disease = disease_x.mean(axis=0)
mean_expression_healthy = healthy_x.mean(axis=0)

# Calculate the absolute difference in mean expression between the two groups
expression_diff = np.abs(mean_expression_disease - mean_expression_healthy)

# Find the indices of the top 10 genes with the largest differences
top_10_diff_genes_indices = np.argsort(expression_diff)[-10:]

# Get the names of the top 10 genes
top_10_diff_genes_names = [f"Gene_{i}" for i in top_10_diff_genes_indices]

# Print the results
print("Top 10 genes with the largest expression differences:", top_10_diff_genes_names)
```
