```python
# Day 3: NumPy and Gene Expression Matrix

## 1. Matrix shape and indexing
# Goal: learn matrix in Numpy
import numpy as np

X = np.array([[2, 5, 8, 1], [3, 6, 7, 2], [9, 1, 2, 8]])

print(f"3 samples (axis 0) × 4 genes (axis 1):", X.shape)

# the first sample's genes
print(f"the first sample's genes:", X[0])

# the second sample's genes
print(f"the second sample's genes:", X[1])

# the first gene's expression across all samples
print(f"the first gene's expression across all samples:", X[:, 0])

# the third gene's expression across all samples
print(f"the third gene's expression across all samples:", X[:, 2])

# the expression of the fourth gene in the second sample
print(f"the expression of the fourth gene in the second sample:", X[1, 3])
```

```python
## 2. Axis operations
print(f"the first two sample:", X[:2])

print(f"the first two genes across all samples", X[:, :2])

print(f"gene 0,1 of sample 1,2", X[:2, :2])

gene_mean = X.mean(axis=0)
print(f"mean expression of each gene across samples", gene_mean)

sample_mean = X.mean(axis=1)
print(f"mean expression of each sample across genes", sample_mean)

gene_sum = X.sum(axis=0)
print(f"sum expression of samples in gene", gene_sum)
sample_sum = X.sum(axis=1)
print(f"sum expression of genes in sample", sample_sum)

gene_max = X.max(axis=0)
print(f"max expression of samples in gene", gene_max)
sample_max = X.max(axis=1)
print(f"max expression of genes in sample", sample_max)

gene_variance = X.var(axis=0)
print(f"variance expression of samples in gene", gene_variance)
sample_variance = X.var(axis=1)
print(f"variance expression of genes in sample", sample_variance)

gene_std = X.std(axis=0)
print(f"std expression of samples in gene", gene_std)
sample_std = X.std(axis=1)
print(f"std expression of genes in sample", sample_std)
```

```python
## 3. Boolean indexing
labels = np.array([
    "control",
    "control",
    "disease"
])

labels == "disease"

disease_X = X[labels == "disease"]
print(f"the disease samples:", disease_X)
print(disease_X.shape)

control_X = X[labels == "control"]
print(f"the control samples:", control_X)
print(control_X.shape)

# why the number of sample of boolean mask have to be identical with the number of sample of matrix? because the boolean mask is used to select specific samples from the matrix. Each element in the boolean mask corresponds to a sample in the matrix, and if the number of elements in the boolean mask does not match the number of samples in the matrix, it would be unclear which samples to select. Therefore, the boolean mask must have the same length as the number of samples in the matrix to ensure that each sample can be properly evaluated for selection.

gene_names = np.array([
    "TP53",
    "EGFR",
    "BRCA1",
    "KRAS"
])

gene_mask = np.array([True, False, True, False])
selected_X = X[:, gene_mask]
print(f"the selected genes:", selected_X)

# tips: X[...] is different with X[...,:] because X[...] is used to select saples, while X[...,:] is used to select genes. The ellipsis (...) is a shorthand notation that represents all dimensions of the array, so X[...] selects all samples, while X[...,:] selects all genes.
```

```python
## 4. Gene filtering
# select genes with mean expression greater than 4
high_expression_mask = gene_mean > 4
highly_expressed_genes = gene_names[high_expression_mask]
highly_expressed_X = X[:, high_expression_mask]

print(f"The highly expressed genes: {highly_expressed_genes}")
print(f"Shape of the filtered expression matrix: {highly_expressed_X.shape}")
```

```python
## 5. Simulated gene-expression dataset
# Goal: simulate a gene-expression dataset with 100 samples and 500 genes, where the first 50 samples are controls and the last 50 samples are disease samples. The disease samples have higher expression in the first 10 genes.

import numpy as np
np.random.seed(42)

n_samples = 100
n_genes = 500

X = np.random.normal(
    loc=5,
    scale=2,
    size=(n_samples, n_genes)
)

labels = np.array(
    ["control"] * 50 +
    ["disease"] * 50
)

gene_names = np.array(
    [f"Gene_{i}" for i in range(n_genes)]
)

X[labels == "disease", 0:10] += 3

print(np.unique(labels, return_counts=True))
```

```python
## 6. Group comparison
# Goal: compare the mean expression of genes between control and disease samples, and identify the top 20 genes with the largest increase in disease expression.

control_X = X[labels == "control"]
disease_X = X[labels == "disease"]
print(f"Control samples shape: {control_X.shape}")
print(f"Disease samples shape: {disease_X.shape}")

control_mean = control_X.mean(axis=0)
disease_mean = disease_X.mean(axis=0)
print(f"Control mean shape: {control_mean.shape}")
print(f"Disease mean shape: {disease_mean.shape}")

expression_diff = disease_mean - control_mean
print(f"Expression difference shape: {expression_diff.shape}")

#the top 20 expression genes with the largest increase in disease expression
top_20_genes_ind = np.argsort(expression_diff)[-20:]
top_20_genes = gene_names[top_20_genes_ind]
top_20_diff = expression_diff[top_20_genes_ind]

print(f"Top 20 genes with the largest increase in disease expression:")
print(top_20_genes)

# tips: np.argsort() returns the indices that would sort an array. By using [-20:], we select the last 20 indices, which correspond to the genes with the largest increase in expression in disease samples compared to control samples.
```

```python
## 7. Highly variable genes
# find top 20 highly variable genes across all samples
gene_variance = X.var(axis=0)
top_20_var_genes_ind = np.argsort(gene_variance)[-20:]
top_20_var_genes = gene_names[top_20_var_genes_ind]
print(f"Top 20 highly variable genes across all samples:")
print(top_20_var_genes)

#top 20 genes with highest disease-control expression difference
top_20_disease_upregulated_ind = np.argsort(disease_mean)[-20:]
top_20_disease_upregulated = gene_names[top_20_disease_upregulated_ind]
print(f"Top 20 highly variable genes across all disease samples:")
print(top_20_disease_upregulated)

#tips: what's the differenth with "highly variable genes" and "disease-upregulated genes"? Highly variable genes are those that show a large amount of variation in expression across all samples, regardless of whether they are control or disease samples. Disease-upregulated genes, on the other hand, are those that have higher expression in disease samples compared to control samples. In other words, highly variable genes may not necessarily be associated with the disease, while disease-upregulated genes are specifically associated with the disease condition. 
```

```python
## 8. Broadcasting and normalization
gene_mean = X.mean(axis=0)
print(gene_mean.shape)
print(X.shape)
X_centered = X - gene_mean
X_centered.mean(axis=0)
print(f"mean expression of genes in sample after centering:")
print(X_centered.mean(axis=0)[0:9])

#tips: Broadcasting is a powerful feature in NumPy that allows for operations between arrays of different shapes. In this case, we are subtracting the mean expression of each gene (a 1D array) from the expression values of all samples (a 2D array). NumPy automatically "broadcasts" the 1D array across the 2D array, allowing for element-wise subtraction without the need for explicit looping. This is useful for normalizing gene expression data by centering it around zero.

#Z-score
X_zscored = (X - gene_mean) / X.std(axis=0)
print(f"mean expression of genes in sample after z-scoring:")
print(X_zscored.mean(axis=0)[0:9])

X_zscored.mean(axis=0)
print(f"mean expression of genes in sample after z-scoring:")
print(X_zscored.mean(axis=0)[0:9])

X_zscored.std(axis=0)
print(f"std expression of genes in sample after z-scoring:")
print(X_zscored.std(axis=0)[0:9])

#tips: Z-score normalization is a common technique used to standardize data. It transforms the data such that the mean of each gene's expression across samples becomes 0 and the standard deviation becomes 1. This is done by subtracting the mean and dividing by the standard deviation for each gene. Z-scoring is particularly useful when comparing genes with different expression scales, as it allows for a fair comparison of their relative expression levels across samples.
```

```python
## 9. Comprehensive exercise
# Goal: simulate a gene-expression dataset with 120 samples and 1000 genes, where the first 60 samples are controls and the last 60 samples are disease samples. The disease samples have higher expression in genes 50 to 64.

np.random.seed(123)

X = np.random.normal(
    5,
    2,
    size=(120, 1000)
)

labels = np.array(
    ["control"] * 60 +
    ["disease"] * 60
)

gene_names = np.array(
    [f"Gene_{i}" for i in range(1000)]
)

X[labels == "disease", 50:65] += 2.5

# the number of samples and genes
n_samples, n_genes = X.shape

# the number of control and disease samples
control_X = X[labels == "control"]
disease_X = X[labels == "disease"]
n_control = control_X.shape[0]
n_disease = disease_X.shape[0]

# mean expression of each gene across samples
gene_mean = X.mean(axis=0)

# the variance of genes of all samples
gene_variance = X.var(axis=0)

# the top 20 highly variable genes across all samples
top_20_var_genes_ind = np.argsort(gene_variance)[-20:]
top_20_var_genes = gene_names[top_20_var_genes_ind]

# the mean expression of genes of control and disease samples
control_mean = control_X.mean(axis=0)
disease_mean = disease_X.mean(axis=0)

# calculate disease - control expression difference
expression_diff = disease_mean - control_mean

# the top 20 genes with highest disease-control expression difference
top_20_disease_upregulated_ind = np.argsort(expression_diff)[-20:]
top_20_disease_upregulated = gene_names[top_20_disease_upregulated_ind]

# find how many top 20 disease-upregulated genes are belong to gene_50 to gene_64
top_20_disease_upregulated_set = set(top_20_disease_upregulated)
gene_50_to_64_set = set(gene_names[50:65])
overlap_genes = top_20_disease_upregulated_set.intersection(gene_50_to_64_set)

# gene-wise Z-score normalization
X_zscored = (X - gene_mean) / X.std(axis=0)

# Final structured output
print(f"Samples: {n_samples}")
print(f"Genes: {n_genes}")
print(f"Control: {n_control}")
print(f"Disease: {n_disease}\n")

print("Top disease-associated genes:")
for gene in top_20_disease_upregulated:
    print(gene)
print()

print(f"Recovered true signal genes: {len(overlap_genes)} / 15")

## Biological interpretation
# Variance identifies genes with large sample-to-sample variation,while differential expression identifies genes whose mean expression differs between groups.
```
